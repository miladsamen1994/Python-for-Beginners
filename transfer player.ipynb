```python
players = []

def add_players():
    players.append("Tiktary")
    players.append("kasra")
    players.append("Jalali")
    players.append("Eydi")
    players.append("Poria")
    players.append("Reza")
    print("_"*20)

     
def show_players():
    for player in players:
        print(player)
    print("Number of players:",len(players))
    print()

def transfer_player():
    players.insert(1,"Amir")
    players.remove("Reza")
    print("Transfer completed!")
    print("_"*20)

def last_player_left():
    last_player = players.pop()
    print("last player left:",last_player)
    print("_"*20)

add_players()

show_players()

transfer_player()

show_players()

last_player_left()
show_players()
 

```
