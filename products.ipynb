```python
prodoucts = []

def add_products():
    prodoucts.append("Laptop")
    prodoucts.append("Mouse")
    prodoucts.append("Keyboard")
    prodoucts.append("Monitor")
    prodoucts.append("Headphone")

def show_products():
    print("=== Product List ===")
    for prodouct in prodoucts:
        print(prodouct)
    print("Number of prodoucts:",len(prodoucts))
    print("_"*20)

def update_products():
    prodoucts.insert(2,"Webcam")
    prodoucts.remove("Mouse")
    prodoucts[3]="Gaming Monitor"
    print("Products updated successfully!")
    print("_"*20)

def sold_product():
    sold = prodoucts.pop()
    print("Sold product:",sold)
    print("_"*20)

add_products()

show_products()

update_products()

show_products()

sold_product()

show_products()
```
