```python
def calculate_total(quantity, price):
    return quantity * price


def check_customer(total):
    if total >= 2000:
        print("Customer Status: Excellent Customer")
    elif total >= 1000:
        print("Customer Status: Good Customer")
    else:
        print("Customer Status: Regular Customer")


# Take information
product_name = input("Enter product name: ")
quantity = int(input("Enter quantity: "))
price = float(input("Enter price per product: "))

# Calculate total price
result = calculate_total(quantity, price)

# Show result
print("\n----- Invoice -----")
print("Product:", product_name)
print("Quantity:", quantity)
print("Price per Product:", price)
print("Total Price:", result)

# Check customer status
check_customer(result)
```
