```python
movies = []

def add_movies():
    movies.append("Interstellar")
    movies.append("Inception")
    movies.append("The Dark Knight")
    movies.append("Fight Club")
    movies.append("Joker")
    print("_"*20)

def show_movies():
    print("===Movies favorite list===")
    for movie in movies:
        print("🎬",movies)
        print("Number of movies:",len(movies))
        print("_"*20)

def update_movies():
    movies.insert(2,"Avatar")
    movies.remove("Joker")
    movies[4]="The Matrix"
    print("Movies updated successfully!")
    print("_"*20)

def favorite_movie():
    print("My favorite movie:",movies[0])
    print("_"*20)

def last_movie_removed():
    last=movies.pop()
    print("Removed movie:",last)
    print("_"*20)

add_movies()

show_movies()

update_movies()

show_movies()

favorite_movie()

last_movie_removed()

show_movies()

```
