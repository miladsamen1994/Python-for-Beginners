```python
players = ["Ali", "Reza", "Hassan", "Amir", "Mohsen"]

def show_players():
    print("=== Team Players ===")

    for i in range(len(players)):
        print(i + 1, "-", players[i])

    print("Number of players:", len(players))
    print("_" * 20)


def show_even_numbers():
    print("=== Even Numbers ===")

    for number in range(2, 11, 2):
        print(number)

    print("_" * 20)


def remove_last_player():
    removed = players.pop()
    print("Removed player:", removed)
    print("_" * 20)


show_players()

show_even_numbers()

remove_last_player()

show_players()
```
