```python
students = ["Ali","Mosi","Saji","Bayram","sina"]

def show_student():
    print("==List of Student==")

    for i in range(len(students)):
        print(i+1,"-",students[i])

    print("Number of students:",len(students))
    print("_"*20)

def show_even_students():
    print("==Even Number==")
    for number in range(2,11,2):
        print(number)

    print("_"*20)

def remove_last_student():
    removed=students.pop()
    print("Removed students:",removed)  
    print("_"*20) 

show_student()  
show_even_students()
remove_last_student()
show_student()
```
