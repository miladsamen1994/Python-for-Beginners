```python
def player_effectiveness(goals, assists, matches):
    return (goals * 4 + assists * 3) / matches


def check_status(score):
    if score >= 8:
        print("Status: World Class")
    elif score >= 5:
        print("Status: Good Player")
    else:
        print("Status: Average Player")


# Take information
player_name = input("Enter player name: ")
goals = int(input("Enter player goals: "))
assists = int(input("Enter player assists: "))
matches = int(input("Enter player matches: "))

# Calculate effectiveness
result = player_effectiveness(goals, assists, matches)

# Show result
print("\n----- Player Report -----")
print("Player Name:", player_name)
print("Score:", result)

# Check status
check_status(result)
```
