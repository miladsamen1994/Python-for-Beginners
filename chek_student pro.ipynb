```python
scores = [95, 83, 74, 61, 45]

def show_scores():
    print("=== Students Scores ===")
    for score in scores:
        print(score)
    print("Number of scores:",len(scores))

    print("_"*20)
def check_grades():
    for score in scores:
        if score >= 90 :
            print(score,"-> Grade A")
        elif score >= 80 :
            print(score,"-> Grade B")
        elif score >= 70 :
            print(score,"-> Grade C")
        else:
            print(score," -> Failed")


    print("_"*20)
def remove_last_score():
    last= scores.pop()
    print("Removed score:",last)
    print("_"*20)

show_scores()

check_grades()

remove_last_score()

show_scores()
```
