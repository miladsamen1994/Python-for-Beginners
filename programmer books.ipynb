```python
books = ["Python", "Java", "C++", "JavaScript", "Go"]
def show_books():
    print("=== Library Books ===")
    for book in range(len(books)):
        print(book+1,"_",books[book])
    print("Number of books:",len(books))
    print("_"*20)
def update_books():
    books.remove("Java")
    books.insert(1,"C#")
    books[4]="Rust"
    print("Books updated successfully!")
    print("_"*20)
def check_books():
    for book in books:
        if book == "Python":
            print(book,"-> Beginner")
        elif book == "Rust":
            print(book,"-> Modern Language")
        else:
            print(book,"> Programming Language")
    print("_"*20)
def remove_last_book():
    removed = books.pop()
    print("Removed book:",removed)

show_books()

update_books()

show_books()

check_books()

remove_last_book()

show_books()                                   
```
