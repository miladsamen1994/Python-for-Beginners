```python
scores = [95, 72, 88, 60, 45, 100]

def show_scores():
    print("=== Students Scores ===")
    for score in scores:
        print("📚 Score:",score)
    print("Number of scores:",len(scores))

    print("_"*20)

def check_scores():
    for score in scores:
        if score >= 80:
            print(score, "-> Passed ✅")
        else:
            print(score, "-> Failed ❌")

    print("_" * 20)

        
def remove_last_score():
    last = scores.pop()
    print("Removed score:",last)

show_scores()

check_scores()

remove_last_score()

show_scores()           
```
