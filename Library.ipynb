```python
books =["شاهنامه","بوستان سعدی","گلستان سعدی","کلیله و دمنه","بوف کور","کباب غاز"]

def show_books():
    print("==لیست کتاب های کتابخانه==")
    print("1-",books[0])
    print("2-",books[1])
    print("3-",books[2])
    print("4-",books[3])
    print("5-",books[4])
    print("6-",books[5])
    print("تعداد کتاب های موجود:",len(books))
    print()

def change_book():
    books[3]= "Python Crash Course"  
    print("به روز رسانی انجام شد!")
    print()

def favorite_book():
    print("کتاب مورد علاقه من:",books[0])
    print()

show_books()
favorite_book()
change_book()
show_books()    
```
